```python
df['TotalCharges']=pd.to_numeric(
    df['TotalCharges'],
    errors='coerce'
)
```

```python
df.isnull().sum()
```

```python
df['TotalCharges'].fillna(
    df['TotalCharges'].median(),
    inplace=True
)
```

```python
df.isnull().sum()
```

```python
df['Churn']=df['Churn'].map({
    'Yes':1,
    'No':0
})
```

```python
df['Partner']=df['Partner'].map({
    'Yes':1,
    'No':0
})
```

```python
df.head()
```

```python
df['gender']=df['gender'].map({
    'Male':1,
    'Female':0
})
```
